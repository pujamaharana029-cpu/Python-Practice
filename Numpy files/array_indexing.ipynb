```python
#2d matrix slicing
arr=np.arange(1,31).reshape(6,5)
arr
```

```python
arr[0]#first element targets rows
```

```python
arr[5,4]#first element targets rows and second element targets columns
```

```python
arr[2:4,2:4]#first element targets rows and second element targets columns
```

```python
slice = arr[3:6,3:5]
slice
```
