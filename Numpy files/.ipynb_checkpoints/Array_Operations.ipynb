```python
import numpy as np
```

Arithematic operations


```python
#Arithematic operations
a1 = np.array([1,2,3,4,5])
a2 = np.array([6,7,8,9,10])
```

```python
arr =a1+a2
arr
```

```python
a1-a2
```

```python
a1*a2
```

```python
a1/a2
```

```python
a1%a2
```

```python
a1//a2
```

Broadcasting

```python
#Broadcasting
l=[10,20,30,40]
arr=np.array(l)
```

```python
arr+10 #Broadcasting used when we have operate each element the same operation
```

```python
arr2 = np.arange(1,21).reshape(4,5)
arr2
```

```python
arr2+10
```

```python
#Swallow copy and deep copy
a=np.arange(1,21)
a
```

```python
slice =a[0:5]
slice*10
```

Matrix operation


```python
#Matrix operation
mat1 = np.arange(1,5).reshape(2,2)
mat2 =np.arange(5,9).reshape(2,2)
```

```python
mat1
```

```python
mat2
```

```python
mat1+mat2
```

```python
mat1@mat2 #multiply matrix
```

```python
np.dot(mat1,mat2)
```

```python
mat1.T
```

Stacking Array

```python
#Stacking Array
a=np.array([1,2,3,4])
b=np.array([5,6,7,8])
```

```python
np.vstack((a,b))
```

```python
np.hstack((a,b))
```

```python
np.column_stack((a,b))
```

Splitting Array

```python
c=np.arange(16).reshape(4,4)
c
```

```python
np.hsplit(c,2) #horizontal split
```

```python
np.vsplit(c,2)
```
