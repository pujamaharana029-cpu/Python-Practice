```python
df=pd.read_csv('poem.csv')
df.head()
```

```python
df.columns
```

```python
#find the number of poems that contain the word "love" in their text
df[df['Poem'].str.contains('love', case=False)].shape[0]
```

```python
#Find the top 5 most frequently occurring words in the poem text column, ignoring:
"""Uppercase/lowercase differences
Punctuation
Empty or missing poem values"""

words = (
    df['Poem']
    .dropna()
    .str.lower()
    .str.replace(r'[^\w\s]', '', regex=True)
    .str.split()
    .explode()
)

words.value_counts().head(5)
```
