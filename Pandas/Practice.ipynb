Series Practice


```python
import numpy as np
import pandas as pd
```

```python
Subject=['Python','Sql','Java','Html','Css']
Marks=[85,78,92,65,88]
```

```python
pd.Series(data=Marks,index=Subject)
```

```python
pd.Series(Marks).mean()
```

```python
pd.Series(Marks).max()
```

```python
pd.Series(Marks).dtype
```

```python
pd.Series(data=Marks,index=Subject).idxmax()  #reurns the index of the maximum value in the series
```

```python
Subject_marks=pd.Series(data=Marks,index=Subject)
Subject_marks[Subject_marks>80]  #returns the values greater than 80
```

```python
Subject_marks[Subject_marks<=90]
```
