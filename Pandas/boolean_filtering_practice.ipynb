```python
import numpy as np
import pandas as pd
```

```python


data = {
    "Name": ["Amit", "Riya", "Puja", "Rahul"],
    "Marks": [85, 92, 67, 78]
}

df = pd.DataFrame(data)

# Students scoring above 80
result = df[df["Marks"] > 80]

print(result)
```

```python
#calculate the average marks
average_marks = df["Marks"].mean()
average_marks
```

```python
#filter students whose marks are greater than the average
result = df[df["Marks"] > average_marks]
result
```

```python
#display their names and marks
result[["Name", "Marks"]]
```

```python
import pandas as pd

data = {
    "Product": ["Laptop", "Phone", "Tablet", "Laptop", "Phone", "Tablet", "Laptop"],
    "Category": ["Electronics", "Electronics", "Electronics", "Electronics", "Electronics", "Electronics", "Electronics"],
    "Price": [60000, 25000, 30000, 55000, 20000, 35000, 70000],
    "Quantity": [2, 5, 3, 4, 6, 2, 1],
    "City": ["Delhi", "Mumbai", "Delhi", "Bangalore", "Mumbai", "Delhi", "Bangalore"]
}

df =pd.DataFrame(data)
df
```

```python
#find all products where price is greter than 30000 and quantity is less than 4 and als0 city is either delhi or bangalore, display product,price,quantity and city columns only
filtered_products = df[(df["Price"] > 30000) & (df["Quantity"] < 4) & (df["City"].isin(["Delhi", "Bangalore"]))][["Product", "Price", "Quantity", "City"]]
filtered_products
```
