```python
import pandas as pd

data = {
    "Employee": ["Amit", "Riya", "Puja", "Rahul", "Sneha", "Karan"],
    "Department": ["IT", "HR", "IT", "Finance", "HR", "IT"],
    "Salary": [35000, 28000, 45000, 50000, 32000, 60000],
    "Experience": [2, 1, 3, 5, 2, 6]
}

df = pd.DataFrame(data)
```

```python
#find the employees who work in the it department
it_employees = df[df["Department"] == "IT"]
it_employees
```

```python
#have a salary greater than 40000
high_salary_employees = df[df["Salary"] > 40000]
high_salary_employees
```

```python
#have more than 2 years of experience
experienced_employees = df[df["Experience"] > 2]
experienced_employees
```
